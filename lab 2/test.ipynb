```python
def fitness_check(capital,historical_prices):
        og_capital=capital
        stop_loss=-2.3
        take_profit=6.2
        trade_size_perct=22
        init_trade_size=round(capital*(trade_size_perct/100),2)
        print(init_trade_size)
        for i in range (len(historical_prices)):
            if historical_prices[i]<stop_loss:
                profit_loss_multiplier=stop_loss/100
                profit_loss=round(init_trade_size*profit_loss_multiplier,2)
                capital+=profit_loss
                init_trade_size=capital*(trade_size_perct/100)
                print('stoploss',capital,init_trade_size)
            elif historical_prices[i]>take_profit:
                profit_loss_multiplier=take_profit/100
                profit_loss=round(init_trade_size*profit_loss_multiplier,2)
                capital+=profit_loss
                init_trade_size=capital*(trade_size_perct/100) 
                print('takeprofit',capital,init_trade_size)   
            else:
                profit_loss_multiplier=historical_prices[i]/100
                print(profit_loss_multiplier)
                profit_loss=round(init_trade_size*profit_loss_multiplier,2)
                capital+=profit_loss
                init_trade_size=capital*(trade_size_perct/100)
                print(capital,init_trade_size)
        
        return round(capital,2)
fitness_check(1000,[-1.2,  3.4,  -0.8,  2.1,  -2.5,  1.7,  -0.3,  5.8,  -1.1,  3.5])
```
