### Why Scaling is Required
Scaling is an essential step in data preprocessing for machine learning models. It ensures that all features contribute equally to the model's performance by bringing them to a similar scale. This is particularly important for algorithms that rely on distance metrics (e.g., K-Nearest Neighbors, Support Vector Machines) or gradient-based optimization (e.g., Logistic Regression, Neural Networks). Without scaling, features with larger ranges can dominate the learning process, leading to biased results.

### Parameters of `train_test_split` in sklearn
The `train_test_split` function in sklearn is used to split a dataset into training and testing subsets. Below are its key parameters:

1. **`arrays`**: The input data to split (e.g., features and target variables).
2. **`test_size`**: Proportion of the dataset to include in the test split (e.g., `0.2` for 20%). Can also be an integer specifying the number of test samples.
3. **`train_size`**: Proportion of the dataset to include in the train split. If not specified, it is automatically set to `1 - test_size`.
4. **`random_state`**: Controls the shuffling applied before splitting. Setting a fixed value ensures reproducibility of results.
5. **`shuffle`**: Whether or not to shuffle the data before splitting. Default is `True`.
6. **`stratify`**: If not `None`, data is split in a stratified fashion based on the provided labels. Useful for maintaining class distribution in both training and testing sets.

```python
from sklearn.model_selection import train_test_split
heart_disease_x=heart_disease.drop(['Heart Disease (in next 10 years)'], axis=1) #features splitted

heart_disease_y=heart_disease['Heart Disease (in next 10 years)'] #target
heart_disease_X_train, heart_disease_X_test, heart_disease_y_train, heart_disease_y_test = train_test_split(heart_disease_x, heart_disease_y,random_state=42) #shuffling interger of 42, its a common convention these days
print(heart_disease_X_test.shape[0]/heart_disease_x.shape[0]*100)
print(heart_disease_X_train.shape[0]/heart_disease_x.shape[0]*100)

```

```python

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaler.fit((heart_disease_X_train))
```

```python
heart_disease_X_train_scaled=scaler.transform(heart_disease_X_train)
```

```python
print("per-feature minimum before scaling:\n {}".format(heart_disease_X_train.min(axis=0)))
print("per-feature maximum before scaling:\n {}".format(heart_disease_X_train.max(axis=0)))
```

```python
print("per-feature minimum after scaling:\n {}".format(heart_disease_X_train_scaled.min(axis=0)))
print("per-feature maximum after scaling:\n {}".format(heart_disease_X_train_scaled.max(axis=0)))
```

```python
# scaler.fit((heart_disease_X_test))
heart_disease_X_test_scaled=scaler.transform(heart_disease_X_test)
```

## Advanced Evaluation- 
**Precision** = What % of predicted positives were *actually* positive?  
  > *"How accurate are my positive predictions?"*

- **Recall** = What % of actual positives were correctly predicted?  
  > *"How well did I find all the real positives?"*

- **F1 Score** = The balance between precision and recall (harmonic mean).  
  > *"How good is the model overall at finding positives accurately?"*

### Formulae:
- Precision = TP / (TP + FP)  
- Recall = TP / (TP + FN)  
- F1 Score = 2 × (Precision × Recall) / (Precision + Recall)

Where:
- TP = True Positives  
- FP = False Positives  
- FN = False Negatives

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
import matplotlib.pyplot as plt

y_pred_log = logistic_regression.predict(heart_disease_X_test_scaled)
# Logistic Regression
target_names = heart_disease['Heart Disease (in next 10 years)'].unique()
cm_log = confusion_matrix(heart_disease_y_test, y_pred_log)
ConfusionMatrixDisplay(cm_log, display_labels=target_names).plot(cmap="Blues")
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

```python
from sklearn.metrics import classification_report

print("Logistic Regression:")
print(classification_report(heart_disease_y_test, y_pred_log, target_names=target_names.astype(str)))
```

```python
from sklearn.metrics import accuracy_score,precision_score, recall_score, f1_score
accuracy = accuracy_score(heart_disease_y_test, y_pred_log)
precision = precision_score(heart_disease_y_test, y_pred_log)
recall = recall_score(heart_disease_y_test, y_pred_log)
f1 = f1_score(heart_disease_y_test, y_pred_log)
print(f"accuracy: {accuracy}")
print(f"precision: {precision}")
print(f"recall: {recall}")
print(f"f1: {f1}")
```

```python
from sklearn.metrics import r2_score, mean_squared_error

# Assuming you have actual y_test and predicted values in predictions
r2 = r2_score(heart_disease_y_test, y_pred_log)
mse = mean_squared_error(heart_disease_y_test, y_pred_log)

print(f"R² Score: {r2:.4f}")
print(f"Mean Squared Error: {mse:.4f}")
```
