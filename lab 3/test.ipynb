```python
import math
import random
class Node:
    def __init__(self):
        self.left=None
        self.right=None
        self.value=None
def strength(x):
    return math.log(x+1,2)+(x/10)
def utility(maxV,minV):
    return strength(maxV)-strength(minV)+(-1)**(random.randint(0,1))*(random.randint(1,10)/10)


def TreeMaker(maxV,minV): 
    stack=[]
    root=Node()
    stack.append(root)
    for i in range(5):
        test=[]
        while len(stack)!=0:
            cur=stack.pop()
            cur.left=Node()
            if i==4:
                cur.left.value=round(utility(maxV,minV),2)
            
            cur.right=Node()
            if i==4:
                cur.right.value=round(utility(maxV,minV),2)
          
            test.append(cur.left)
            test.append(cur.right)
            
        stack=stack+test
    return root
```

```python
class Node:
    def __init__(self):
        self.left=None
        self.right=None
        self.value=None
stack=[]
root=Node()
stack.append(root)
value=1
for i in range(5):
    test=[]
    while len(stack)!=0:
        cur=stack.pop()
        cur.left=Node()
        if i==4:
            cur.left.value=value
         
        cur.right=Node()
        if i==4:
            cur.right.value=value
            value+=1
        test.append(cur.left)
        test.append(cur.right)
        
    stack.extend(test)
    
     
def print_tree(root):
    # If the tree is empty
    if not root:
        print("Tree is empty.")
        return
    
    # Queue for level-order traversal
    queue = [root]
    
    while queue:
        # Number of nodes at the current level
        level_length = len(queue)
        level_values = []  # To store node values at the current level
        
        for _ in range(level_length):
            node = queue.pop(0)  # Dequeue the node
            if node:
                level_values.append(str(node.value))  # Mark nodes as "X" for visualization
                # Only add children to the queue if they exist
                if node.left or node.right:
                    queue.append(node.left)
                    queue.append(node.right)
            else:
                level_values.append("None")  # Placeholder for None nodes
        
        # Print the level if there are any nodes at this level
        if level_values:
            print(" ".join(level_values))
print_tree(root)
```
